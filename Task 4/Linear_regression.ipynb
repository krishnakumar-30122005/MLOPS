```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt

df = pd.read_csv('canada.csv')
print(df)


year = df[['year']]
income = df['per capita income (US$)']

model = linear_model.LinearRegression()
model.fit(year,income)

prediction = model.predict([[2020]])
print("Prediction ===> ", prediction[0])
```

# Model storing using Pickel

```python
import pickle
with open("Model1","wb") as file:
    pickle.dump(model,file)
with open("Model1","rb") as file:
    modell = pickle.load(file)
print(modell.predict([[2020]]))
```

# Model stroing using Joblib

```python
import joblib
joblib.dump(model,"Model2")
modelll = joblib.load("Model2")
print(modelll.predict([[2020]]))
```
