```python
import pandas as pd
df = pd.read_csv("titanic.csv")

# print(df)
# print(df.head())
# print(df.info())
df["Age"] = df["Age"].fillna(df["Age"].mean())
print(df.isnull().sum())

# print(df["Cabin"])
```

```python
result_df = df[["Pclass", "Sex", "Age", "Fare"]]
print(result_df)
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
result_df["Sex"] = encoder.fit_transform(df["Sex"])
# print(result_df["Sex"])
```

```python
X = result_df
Y = df["Survived"]


from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()

model.fit(
    X,
    Y
)
model.score(X,Y)
```
